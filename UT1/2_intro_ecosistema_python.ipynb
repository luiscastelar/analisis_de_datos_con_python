## 2. Introducción al ecosistema de datos en Python
* **Fundamentos de rendimiento y estructuras:**
  * Diferencias entre colecciones nativas de Python (`list`, `dict`) y estructuras optimizadas para análisis masivo.
  * Introducción a **NumPy**: Arrays multidimensionales (`ndarray`), vectorización y operaciones lógicas/aritméticas sin bucles explícitos (`for`).
* **Estructuras fundamentales en Pandas:**
  * **Series:** Estructura unidimensional con etiquetas asociadas (índice). Creación, acceso por posición e índice, y operaciones vectorizadas básicas.
  * **DataFrame:** Tabla bidimensional con filas y columnas etiquetadas. Anatomía de un DataFrame (índice, columnas, valores).
* **Selección e indexación básica:**
  * Indexación basada en etiquetas con `.loc[]`.
  * Indexación basada en posición entera con `.iloc[]`.
  * Filtrado booleano (máscaras booleanas) para la selección condicional de datos.
