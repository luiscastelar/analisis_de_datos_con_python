## 1. Captura de datos desde CSV (CE1.a, CE1.b, CE1.c)

El formato **CSV** (*Comma Separated Values*) es texto plano: un registro por línea y un separador entre columnas.
Es el formato de intercambio más extendido, pero no guarda tipos: todo entra como texto y Pandas debe inferirlos.

Pandas lee con `pd.read_csv()` y escribe con `DataFrame.to_csv()`.

**Parámetros de `read_csv()` que conviene controlar:**
- `sep`: separador de campos (`","` por defecto; `";"` es lo habitual en España).
- `encoding`: codificación del fichero (`"utf-8"`, `"latin-1"`).
- `dtype`: tipo de cada columna, para evitar inferencias no deseadas (por ejemplo códigos postales con ceros a la izquierda).
- `usecols`: leer solo las columnas necesarias (ahorra memoria).
- `na_values`: cadenas que deben interpretarse como nulo (`""`, `"N/A"`, `"-"`).
- `parse_dates`: columnas que deben convertirse a `datetime`.
- `nrows` / `chunksize`: lectura parcial o por bloques para ficheros grandes.

**Parámetros de `to_csv()` que conviene controlar:**
- `index=False`: no escribir el índice como columna (casi siempre lo que queremos).
- `sep`, `encoding`, `decimal`: para adaptar la salida al destinatario.

### 1.1. Ejemplo resuelto: lectura y escritura de un CSV

Creamos un CSV de ejemplo con separador `;` y lo volvemos a leer indicando separador y codificación.

```python
import pathlib
import pandas as pd

pathlib.Path("data_sample").mkdir(exist_ok=True)
ruta_csv = "data_sample/ejemplo_ventas.csv"

# DataFrame -> CSV
pd.DataFrame({
    "id": [1, 2, 3, 4],
    "producto": ["Laptop", "Ratón", "Teclado", "Monitor"],
    "precio": [850.0, 25.5, 45.0, 300.0],
}).to_csv(ruta_csv, sep=";", index=False, encoding="utf-8")

# CSV -> DataFrame
df = pd.read_csv(ruta_csv, sep=";", encoding="utf-8")
print(df)
print(df.dtypes)
```

### 1.2. Ejemplo resuelto: control de tipos, columnas y nulos

`usecols`, `dtype` y `na_values` permiten decidir exactamente qué entra en el DataFrame y con qué tipo.

```python
df_tipos = pd.read_csv(
    ruta_csv,
    sep=";",
    usecols=["id", "precio"],
    dtype={"id": "int32", "precio": "float64"},
    na_values=["", "N/A"],
)

print(df_tipos)
print(df_tipos.dtypes)
```
